```python
from pyspark.sql import functions as F
```

```python
# Read cust and orders dataframes

cust_df = spark.table("samples.tpch.customer")
orders_df = spark.table("samples.tpch.orders")

cust_df.show(5)
orders_df.show(5)
```

```python
# Normal Join

normal_join = (
    cust_df
    .join(
        orders_df,
        on=cust_df.c_custkey == orders_df.o_custkey,
        how="left"
    )
)

display(normal_join)
```

```python
cust_df.count()
```

```python
orders_df.count()
```

```python
# Broadcast Join

broadcast_join = (
    orders_df
    .join(
        F.broadcast(cust_df),
        on=cust_df.c_custkey == orders_df.o_custkey,
        how="inner"
    )
)

display(broadcast_join)
```

```python
%sql

create database rainbow;
use rainbow;
```

```python
broadcast_join.write.mode("overwrite").saveAsTable("customer_orders")
```
