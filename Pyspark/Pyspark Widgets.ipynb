> ### Text Widget

```python
dbutils.widgets.text('TableName', 'samples.nyctaxi.trips')
```

```python
target_table = dbutils.widgets.get('TableName')
print(target_table)
```

> ### Drop Down Widget

```python
dbutils.widgets.dropdown('DropDown', 'employee', ['employee', 'department', 'salary'])
```

```python
dropdown_values = dbutils.widgets.get('DropDown')
print(dropdown_values)
```

> ### Combo Box Widget

```python
dbutils.widgets.combobox('ComboBox', 'employee', ['employee', 'department', 'salary'])
```

```python
combox_values = dbutils.widgets.get('ComboBox')
print(combox_values)
```

> ### Multi-Select Widget

```python
dbutils.widgets.multiselect('MultiSelect', 'employee', ['employee', 'department', 'salary'])
```

```python
multiselect_values = dbutils.widgets.get('MultiSelect')
print(multiselect_values)
```
