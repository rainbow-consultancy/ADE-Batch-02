```python
1. txt
2. csv (comma seperated values) / tsv (tab seperated values) / psv (pipe seperated values)

3. parquet files (par-k) column oriented
4. avro files (row based)

5. Delta Lake House  --> delta format (parquet)  --> ACID (atomicity, consistency, isolation, durability)
```

```python
iris_df = spark.read.format("parquet").load("/mnt/rainbowcontainer/raw-datasets/iris.parquet")
display(iris_df)
```

```python
iris_df.write.format("parquet").mode("overwrite").save("/mnt/rainbowcontainer/Events/iris")
```

```python
iris_df.write.format("delta").mode("overwrite").option("mergeSchema", True).save("/mnt/rainbowcontainer/Events/iris_delta")
```

```python
iris_filtered = iris_df.filter(iris_df["variety"] == "Setosa")
display(iris_filtered)
```

```python
iris_filtered.write.format("delta").mode("overwrite").save("/mnt/rainbowcontainer/Events/iris_delta")
```

```python
iris_new = spark.read.format("delta").load("/mnt/rainbowcontainer/Events/iris_delta")
display(iris_new)
```
