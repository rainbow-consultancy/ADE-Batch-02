```python
from pyspark.sql import functions as F
```

```python
df = spark.table("samples.nyctaxi.trips")
df.display()
```

```python
df2 = df.withColumn("pickup_date", F.date_format("tpep_pickup_datetime", "yyyy-MM-dd"))
df2.groupBy("pickup_date").count().display()
```

```python
df.groupBy("trip_distance").count().display()
```

```python
df.groupBy("pickup_zip").count().display()
```

```python
(
  df
  .write
  .format("parquet")
  .mode("overwrite")
  .partitionBy("tpep_pickup_datetime")  # LOW-CARDINALITY
  .save("/mnt/rainbowcontainer/dataprep/taxi-dataset")
)
```

```python
df2.count()
```

```python
(
  df2
  .write
  .format("parquet")
  .mode("overwrite")
  .partitionBy("pickup_date")  # HIGH-CARDINALITY
  .save("/mnt/rainbowcontainer/dataprep/taxi-dataset")
)
```

```python
df2 = df2.repartition(1)  # full - shuffel (used to both incre or decre no of partitions)
```

```python
df2 = df2.coalesce(1)  # less/no - shuffel (used to decre no of partitions)
```
